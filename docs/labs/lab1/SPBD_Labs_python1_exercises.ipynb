##1. Word Frequency

Create a notebook program that counts the frequency of each word in a text file (for example, the novel "Os Maias").

a) No sorting required: words and their frequencies can appear in any order.

b) Sorted by word, in ascending alphabetical order.

c) Sorted by frequency, in descending order (most frequent words first).


```python
#@title Download Dataset
!wget -q -O os_maias.txt https://www.dropbox.com/s/n24v0z7y79np319/os_maias.txt?dl=0
```

```python
import string
from collections import Counter

# Descarregar o ficheiro (se ainda não o descarregou no notebook)
!wget -q -O os_maias.txt https://www.dropbox.com/s/n24v0z7y79np319/os_maias.txt?dl=0

# Ler e limpar o ficheiro
with open("os_maias.txt", "r", encoding="utf-8") as f:
    text = f.read().lower()

# Remover pontuação do texto
translator = str.maketrans("", "", string.punctuation)
text_cleaned = text.translate(translator)

# Contar a frequência de cada palavra
words = text_cleaned.split()
word_counts = Counter(words)


#a) Mostra as palavras e frequências sem ordem específica

for word, count in word_counts.items():
    print(f"{word}: {count}")
```

```python
#b) Ordenação alfabética pelas chaves (palavras)

sorted_by_word = sorted(word_counts.items(), key=lambda x: x[0])

for word, count in sorted_by_word:
    print(f"{word}: {count}")
```

```python
#c) Ordenação pelo valor da frequência (em ordem decrescente)

sorted_by_frequency = sorted(word_counts.items(), key=lambda x: x[1], reverse=True)

for word, count in sorted_by_frequency:
    print(f"{word}: {count}")
```
